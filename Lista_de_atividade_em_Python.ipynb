```python
print("pratica esportiva: 12 a 17 anos e autorização")

idade = int(input ("idade: "))

autorizacao = input ("possui autorização? ").lower() == "sim"

autorizacao = False

if idade >= 12 and idade <= 17 and autorizacao == True:
  print("autorizado")
else:
    print("não autorizado")
```

```python
chovendo = input("Está chovendo? sim/não ou yes/no(t)?"). lower() == "sim"

if chovendo == True:
  print("Está chovendo, use o guarda-chuva.")
else:
  print("Está fazendo sol.")
```

```python
print("Classificação de criança, adolescente, adulto e idoso")

idade = int (input("qual é sua idade?"))

if idade >= 0 and idade < 12:
  print("criança")
elif idade > 11 and idade < 18:
  print("adolescente")
elif idade > 17 and idade < 59:
  print("adulto")
elif idade > 59:
  print("idoso")
```
